Importing dependies

```python
import numpy as np
```

```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
from sklearn.metrics import accuracy_score
```

Data Collection and preprocessing

```python
# loading data from csv file to a pandas dataframe

raw_mail_data = pd.read_csv('/content/mail_data.csv')
```

```python
print(raw_mail_data)
```

```python
#replace the null values with null string

mail_data = raw_mail_data.where((pd.notnull(raw_mail_data)),'')
```

```python
mail_data.head()
```

```python
#checking row and col

mail_data.shape
```

Label encoding

```python
#label spam mail as 0 non spam mil(ham) as 1

mail_data.loc[mail_data['Category'] == 'spam', 'Category',]=0;
mail_data.loc[mail_data['Category'] == 'ham','Category']=1;
```

spam - 0
ham -1

```python
#seperating the data text and label

x = mail_data['Message']
y = mail_data['Category']
```

```python
print(x)
```

```python
print(y)
```

Splitting the data into traing data & testing data.

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, train_size = 0.2, random_state = 3)
```

```python
print(x.shape)
```

```python
print(x_train.shape)
```

```python
print(x_test.shape)
```

```python
print(x_train)
```

```python
print(x_test)
```

Feature_extraction

```python
#transfrom the test data into feature vector data that can be used to logistic regression

feature_extraction =TfidfVectorizer(min_df =1, stop_words='english')

x_train_features = feature_extraction.fit_transform(x_train)
x_test_features = feature_extraction.transform(x_test)

#convert y_train and y_test values of integer

y_train = y_train.astype('int')
y_test = y_test.astype('int')
```

```python
print(x_train_features)
```

```python
print(x_test_features)
```

Training the model

```python
model = LogisticRegression()
```

```python
# training the logistic model with the training data

model.fit(x_train_features, y_train)
```

Evaluting model

```python
# prediction the training data

prediction_on_training = model.predict(x_train_features)
accuracy_on_training = accuracy_score(y_train, prediction_on_training)
```

```python
print('Accuracy on training data:', accuracy_on_training)
```

```python
#prediction on test data

prediction_on_test_data = model.predict(x_test_features)
```

```python
accuracy_on_test_data = accuracy_score(y_test, prediction_on_test_data)
```

```python
print("Accuracy on test data", accuracy_on_test_data)
```

Building the predicting system

```python
input_mail = ["URGENT! You have won a 1 week FREE membership in our £100,000 Prize Jackpot! Txt the word: CLAIM to No: 81010 T&C www.dbuk.net LCCLTD POBOX 4403LDNW1A7RW18"]
```

```python
#convert test to features vectors

input_data_features = feature_extraction.transform(input_mail)
```

```python
#making prediction

prediction = model.predict(input_data_features)

if prediction[0]==1:
  print("Ham mail")
else:
  print("Spam mail")
```
